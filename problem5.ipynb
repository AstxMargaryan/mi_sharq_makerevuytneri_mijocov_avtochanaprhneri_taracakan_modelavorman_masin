Դիցուք տարածության մեջ տրված է բնական հավասարումով տրված

$$
\vec{r}(s)=\{s;\;0;\;0,01s^2\}
$$

կորը, որտեղ $ s \in [0;100]$ մետր։ Ճանապարհի կես լայնությունը՝ $ a=5 \text{ մ} $ իսկ լայնական թեքությունը՝ 0,02(այսինքն 2%)։

Այդ դեպքում մակերևույթի պարամետրական հավասարումը կլինի՝

$$
R(s,t)=\{s;\;t;\;0,01s^2+0,02t\},
\qquad s\in[0;100], \quad t\in[-5;5]
$$

Տուփաձևությունը արտահայտվում է

$$
D=[0;100]\times[-5;5]
$$

պարամետրերի ուղղանկյուն տիրույթով։

```python
import numpy as np
import plotly.graph_objects as go


s_vals = np.linspace(0, 100, 150)
t_vals = np.linspace(-5, 5, 60)

S, T = np.meshgrid(s_vals, t_vals)

X = S
Y = T
Z = 0.01 * S**2 + 0.02 * T


fig = go.Figure()

fig.add_surface(
    x=X,
    y=Y,
    z=Z,
    opacity=0.9
)


s_curve = np.linspace(0, 100, 300)
x_curve = s_curve
y_curve = np.zeros_like(s_curve)
z_curve = 0.01 * s_curve**2

fig.add_trace(
    go.Scatter3d(
        x=x_curve,
        y=y_curve,
        z=z_curve,
        mode="lines",
        line=dict(width=6, color="red"),
        name="r(s)"
    )
)

fig.update_layout(
    title="Ճանապարհի մակերևույթը",
    scene=dict(
        xaxis_title="x = s",
        yaxis_title="y = t",
        zaxis_title="z = 0.01s² + 0.02t",
        aspectmode="manual",
        aspectratio=dict(x=2.5, y=1, z=0.8)
    ),
    width=950,
    height=700
)

fig.show()
```
