Տարածության մեջ տրված է` 

$$
\vec r(u)=\left(u^2+1\right)\vec i+\left(u^3-u_0^3\right)\vec j+\left(2u+\left(5-2u_0\right)\right)\vec k,
\qquad 0.2\le u\le 3,
$$

ուղղորդ կորը որտեղ

$$
u_0=\sqrt{5.5},
$$

որը ընտրված է այնպես, որ

$$
\vec r(u_0)=(6.5,\,0,\,5):
$$

Լայնական հատույթը \(T\)-աձև պատկեր է, և (6.5,0,5)
կետը համապատասխանում է լայնական հատույթի ընտրված կենտրոնին։

Տուփաձև մակերևույթը բաղկացած է ութ նիստերից, այսինքն $ k=1,2,3,4,5,6,7,8:$

Յուրաքանչյուր նիստի պարամետրական հավասարումը տրվում է հետևյալ կերպ`

$$
\vec{\rho}_k(u,v)=
\vec r(u)+
\frac{d_k}{\cos v}
\left(
\cos(\beta_k+v)\,\vec e_0(u)+
\sin(\beta_k+v)\,\vec g_0(u)
\right),
$$

որտեղ $\beta_k$-ն սահմանում է k-րդ կողմի դիրքը հարթության մեջ հատույթի կենտրոնի նկատմամբ։

Մասնավորապես`

- եթե կողմը կենտրոնից վերև է, ապա
  $$
  \beta_k=\frac{\pi}{2},
  $$
- եթե աջ է, ապա
  $$
  \beta_k=0,
  $$
- եթե ներքև է, ապա
  $$
  \beta_k=-\frac{\pi}{2},
  $$
- եթե ձախ է, ապա
  $$
  \beta_k=\pi:
  $$

Ուստի տվյալ \(T\)-աձև  պատկերի համար ստացվում է`

$$
\beta_1=\frac{\pi}{2},\qquad
\beta_2=0,\qquad
\beta_3=\frac{\pi}{2},\qquad
\beta_4=0,
$$

$$
\beta_5=-\frac{\pi}{2},\qquad
\beta_6=\pi,\qquad
\beta_7=\frac{\pi}{2},\qquad
\beta_8=\pi:
$$


իսկ $d_k$-ն կենտրոնի և տվյալ կողմի կոորդինատի տարբերության բացարձակ արժեքն է՝
$$
d_k = |z_{\text{side}} - z_c|
\quad \text{կամ} \quad
d_k = |x_{\text{side}} - x_c|:
$$


Քանի որ \(T\)-աձև  պատկերի գագաթներն են`

$$
(2,8),\ (11,8),\ (11,6),\ (8,6),\ (8,2),\ (5,2),\ (5,6),\ (2,6),
$$

իսկ կենտրոնը`

$$
(x_c,z_c)=(6.5,5),
$$

ստացվում են հետևյալ արժեքները`

$$
d_1=3,\qquad
d_2=4.5,\qquad
d_3=1,\qquad
d_4=1.5,
$$

$$
d_5=3,\qquad
d_6=1.5,\qquad
d_7=1,\qquad
d_8=4.5:
$$

Այստեղ $\vec e_0(u)$ և $\vec g_0(u)$ վեկտորները որոշվում են ուղղորդ կորի Ֆրենեի եռանիստի միջոցով և հաշվարկվում են հետևյալ բանաձևերով`

$$
\vec e_0(u)=
\frac{\left(\vec r\,'(u)\times \vec r\,''(u)\right)\times \vec r\,'(u)}
{\left|\vec r\,'(u)\times \vec r\,''(u)\right|\cdot \left|\vec r\,'(u)\right|},
$$

$$
\vec g_0(u)=
\frac{\vec r\,'(u)\times \vec r\,''(u)}
{\left|\vec r\,'(u)\times \vec r\,''(u)\right|}:
$$


Տուփաձև մակերևույթի յուրաքանչյուր նիստի համար պարամետր $v$-ն սահմանափակվում է որոշակի միջակայքով՝

$$
v \in 
\left[
\arctan\left(\frac{z_1-z_c}{x_1-x_c}\right),
\arctan\left(\frac{z_2-z_c}{x_2-x_c}\right)
\right]:
$$

$$
\begin{aligned}
v &\in \left[-\arctan\frac{3}{2},\,\arctan\frac{3}{2}\right],\ k=1
\qquad
v \in \left[\arctan\frac{2}{9},\,\arctan\frac{2}{3}\right],\ k=2
\\[6pt]
v &\in \left[\arctan\frac{3}{2},\,\arctan\frac{9}{2}\right],\ k=3
\qquad
v \in \left[-\arctan 2,\,\arctan\frac{2}{3}\right],\ k=4
\\[6pt]
v &\in \left[-\arctan\frac{1}{2},\,\arctan\frac{1}{2}\right],\ k=5
\qquad
v \in \left[-\arctan\frac{2}{3},\,\arctan 2\right],\ k=6
\\[6pt]
v &\in \left[-\arctan\frac{9}{2},\,-\arctan\frac{3}{2}\right],\ k=7
\qquad
v \in \left[-\arctan\frac{2}{3},\,-\arctan\frac{2}{9}\right],\ k=8
\end{aligned}
$$

Պահանջվում է կառուցել տուփաձև մակերևույթի 

$$
\rho_k(u,v), \quad k=1,\dots,8
$$
նիստերի պարամետրական հավասարումները, ինչպես նաև կառուցել տուփաձև մակերևույթը որպես այդ նիստերի միավորում՝

$$
S = \bigcup_{k=1}^{8} \rho_k(u,v)
$$

և իրականացնել դրանց գրաֆիկական պատկերումը։



```python
import numpy as np
import sympy as sp
import plotly.graph_objects as go
```

```python
u, v = sp.symbols('u v', real=True)
u0 = sp.sqrt(sp.Rational(11, 2))
r_u = sp.Matrix([
    u**2 + 1,
    u**3 - u0**3,
    2*u + (5 - 2*u0)
])

r_prime = sp.diff(r_u, u)
r_double_prime = sp.diff(r_prime, u)
```

```python
cross_vec = r_prime.cross(r_double_prime)
norm_cross = sp.sqrt(cross_vec.dot(cross_vec))
g0_u = sp.simplify(cross_vec / norm_cross)

numerator_e0 = cross_vec.cross(r_prime)
norm_r_prime = sp.sqrt(r_prime.dot(r_prime))
e0_u = sp.simplify(numerator_e0 / (norm_cross * norm_r_prime))
```

```python
faces_data = {
    1: {"d": 3,                 "beta":  sp.pi/2, "vmin": -sp.atan(sp.Rational(3, 2)), "vmax":  sp.atan(sp.Rational(3, 2))},
    2: {"d": sp.Rational(9, 2), "beta":  0,       "vmin":  sp.atan(sp.Rational(2, 9)), "vmax":  sp.atan(sp.Rational(2, 3))},
    3: {"d": 1,                 "beta":  sp.pi/2, "vmin":  sp.atan(sp.Rational(3, 2)), "vmax":  sp.atan(sp.Rational(9, 2))},
    4: {"d": sp.Rational(3, 2), "beta":  0,       "vmin": -sp.atan(2),                  "vmax":  sp.atan(sp.Rational(2, 3))},
    5: {"d": 3,                 "beta": -sp.pi/2, "vmin": -sp.atan(sp.Rational(1, 2)), "vmax":  sp.atan(sp.Rational(1, 2))},
    6: {"d": sp.Rational(3, 2), "beta":  sp.pi,   "vmin": -sp.atan(sp.Rational(2, 3)), "vmax":  sp.atan(2)},
    7: {"d": 1,                 "beta":  sp.pi/2, "vmin": -sp.atan(sp.Rational(9, 2)), "vmax": -sp.atan(sp.Rational(3, 2))},
    8: {"d": sp.Rational(9, 2), "beta":  sp.pi,   "vmin": -sp.atan(sp.Rational(2, 3)), "vmax": -sp.atan(sp.Rational(2, 9))}
}
```

```python
def rho_expr(k):
    d = faces_data[k]["d"]
    beta = faces_data[k]["beta"]
    return sp.simplify(
        r_u + (d / sp.cos(v)) * (
            sp.cos(beta + v) * e0_u +
            sp.sin(beta + v) * g0_u
        )
    )

rho_list = [rho_expr(k) for k in range(1, 9)]
```

```python
print("r(u) =")
sp.pprint(r_u)

print("\nr'(u) =")
sp.pprint(r_prime)

print("\nr''(u) =")
sp.pprint(r_double_prime)

print("\ng₀(u) =")
sp.pprint(g0_u)

print("\ne₀(u) =")
sp.pprint(e0_u)

for i, rho in enumerate(rho_list, start=1):
    print(f"\nρ_{i}(u,v) =")
    sp.pprint(rho)
```

```python
r_func = sp.lambdify(u, r_u, "numpy")

rho_funcs = [
    (
        sp.lambdify((u, v), rho_k[0], "numpy"),
        sp.lambdify((u, v), rho_k[1], "numpy"),
        sp.lambdify((u, v), rho_k[2], "numpy"),
    )
    for rho_k in rho_list
]
```

```python
def eval_surface(rho_func, U, V):
    fx, fy, fz = rho_func
    X = np.asarray(fx(U, V), dtype=float)
    Y = np.asarray(fy(U, V), dtype=float)
    Z = np.asarray(fz(U, V), dtype=float)
    return np.stack([X, Y, Z], axis=-1)

def add_curve(fig):
    fig.add_trace(go.Scatter3d(
        x=curve_x, y=curve_y, z=curve_z,
        mode='lines',
        line=dict(color='black', width=10),
        name='r(u)'
    ))
    
    fig.add_trace(go.Scatter3d(
        x=[6.5], y=[0], z=[5],
        mode='markers+text',
        marker=dict(size=6, color='red'),
        text=['(6.5, 0, 5)'],
        textposition='top center',
        name='special point'
    ))

def add_surface(fig, S, name):
    fig.add_trace(go.Surface(
        x=S[:, :, 0],
        y=S[:, :, 1],
        z=S[:, :, 2],
        colorscale='Blues',
        showscale=False,
        opacity=0.85,
        name=name
    ))
```

```python
u_min, u_max = 0.2, 3.0
Nu, Nv = 180, 80

u_vals = np.linspace(u_min, u_max, Nu)

u_curve = np.linspace(u_min, u_max, 400)
curve = np.array(r_func(u_curve), dtype=float).reshape(3, -1).T
curve_x, curve_y, curve_z = curve[:, 0], curve[:, 1], curve[:, 2]

surfaces = []

for k in range(1, 9):
    vmin = float(sp.N(faces_data[k]["vmin"]))
    vmax = float(sp.N(faces_data[k]["vmax"]))
    v_vals = np.linspace(vmin, vmax, Nv)
    U, V = np.meshgrid(u_vals, v_vals, indexing='ij')
    S = eval_surface(rho_funcs[k - 1], U, V)
    surfaces.append(S)
```

```python
for i, S in enumerate(surfaces, start=1):
    fig = go.Figure()
    add_surface(fig, S, f'ρ_{i}(u,v)')
    add_curve(fig)
    
    fig.update_layout(
        title=f'ρ_{i}(u,v)',
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Z',
            aspectmode='data'
        ),
        width=900,
        height=700
    )
    
    fig.show()
```

```python
fig_all = go.Figure()

for i, S in enumerate(surfaces, start=1):
    add_surface(fig_all, S, f'ρ_{i}')

add_curve(fig_all)

fig_all.update_layout(
    title='S = ρ₁ ∪ ρ₂ ∪ ρ₃ ∪ ρ₄ ∪ ρ₅ ∪ ρ₆ ∪ ρ₇ ∪ ρ₈',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='data'
    ),
    width=1100,
    height=850
)

fig_all.show()
```
